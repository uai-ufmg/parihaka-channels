```python
import glob
import os
import csv

# TODO: set as file params
base_dir = "/pgeoprj/godeep/dados/l1_datasets"
dataset_dir = "canais_sinteticos_horizontes"
# dataset_dir = "parihaka/processed_images"
dataset_name = "parihaka"
out_file = "parihaka_paths.csv"
is_synth = True
exclude_ds = None
include_ds = None
```

```python
def get_data_paths(data_path, dataset_name):
    paths = []

    # get only seismic data, bc we just need one per horizon
    for path in glob.glob(os.path.join(data_path, "*_seismic.npy")):
        rel_path = path.split(base_dir)[1]
        if rel_path[0] == '/':
            rel_path = rel_path[1:]
        # get the path to both seismic and label
        rel_path = path.split('_seismic')[0]
        paths.append((dataset_name, rel_path))
    
    return paths
```

```python
def is_ds_allowed(ds_name):
    ds = ds_name.split('-')[0]

    if include_ds is not None:
        if ds not in include_ds:
            return False

    if exclude_ds is not None:
        if ds in exclude_ds:
            return False

    return True
```

```python
def get_data_paths_from_dirs(root):
    paths = []

    for path in glob.glob(os.path.join(root, "canal_*")):
        ds_name = path.split('canal_')[1]

        if is_ds_allowed(ds_name):
            ds_paths = get_data_paths(os.path.join(path, "images"), ds_name)
            paths.extend(ds_paths)
    
    return paths
```

```python
def write_to_file(out_file_path, data, header=('dataset_name','relative_path')):
    with open(out_file_path, 'w') as f:
        csv_writer = csv.writer(f)
        csv_writer.writerow(header)
        csv_writer.writerows(data)
```

```python
full_path = os.path.join(base_dir, dataset_dir)

if is_synth:
    data_paths = get_data_paths_from_dirs(full_path)
else:
    data_paths = get_data_paths(full_path, dataset_name)

write_to_file(out_file, data_paths)
```
